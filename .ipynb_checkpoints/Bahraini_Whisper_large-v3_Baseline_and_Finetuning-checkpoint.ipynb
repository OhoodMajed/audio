# Bahraini Whisper Model (large-v3) — Base Test + LoRA Fine-tuning

**Part 1:** Evaluate stock `openai/whisper-large-v3` on Bahraini test clips (baseline).

**Part 2:** Fine-tune the model with LoRA on Bahraini training data and re-evaluate.

## Part 1 — Base Whisper v3 (large) Evaluation (Bahraini)

Goal: measure the stock `openai/whisper-large-v3` on the Bahraini test clips.

Flow: install libraries → imports → verify GPU/CUDA → load dataset → prepare features → load base model → evaluate on test set (WER/CER).

```
import torch
from datasets import load_from_disk, Audio
from huggingface_hub import snapshot_download
from torch.utils.data import DataLoader
from tqdm.auto import tqdm
import evaluate
from transformers import (
    WhisperProcessor,
    WhisperForConditionalGeneration,
    BitsAndBytesConfig
)
```

```
print("GPU:", torch.cuda.get_device_name(0))
print("CUDA available:", torch.cuda.is_available())
```

```
# Download and load the full Bahraini speech dataset
repo_id = "Hishambarakat/Bahraini_Speech_Dataset"
local_path = snapshot_download(
    repo_id=repo_id,
    repo_type="dataset",
    allow_patterns=["dataset_dict.json", "dataset_info.json", "train/**", "test/**"]
)

dataset = load_from_disk(local_path)
dataset = dataset.cast_column("audio", Audio(sampling_rate=16000))
print("Train clips:", len(dataset["train"]))
print("Test clips:", len(dataset["test"]))
```

```
# Filter out empty text and audio clips shorter than 0.5s (8000 samples at 16kHz)
def is_valid(sample):
    valid_text = sample["text"] is not None and len(sample["text"].strip()) > 0
    valid_audio = len(sample["audio"]["array"]) >= 8000
    return valid_text and valid_audio

dataset = dataset.filter(is_valid)
print("Cleaned Train:", len(dataset["train"]))
print("Cleaned Test:", len(dataset["test"]))
```

```
model_id = "openai/whisper-large-v3"
processor = WhisperProcessor.from_pretrained(
    model_id,
    language="Arabic",
    task="transcribe"
)
```

```
def prepare_dataset(batch):
    audio = batch["audio"]
    # Extract audio features
    batch["input_features"] = processor.feature_extractor(
        audio["array"],
        sampling_rate=audio["sampling_rate"]
    ).input_features[0]
    # Tokenize target text
    batch["labels"] = processor.tokenizer(
        batch["text"],
        truncation=True,
        max_length=448
    ).input_ids
    return batch

# Select subsets: 5000 train, 500 test
train_sub = dataset["train"].select(range(5000))
test_sub = dataset["test"].select(range(500))

train_processed = train_sub.map(prepare_dataset, remove_columns=dataset["train"].column_names, num_proc=2)
test_processed = test_sub.map(prepare_dataset, remove_columns=dataset["test"].column_names, num_proc=2)

print("Train samples ready:", len(train_processed))
print("Test samples ready:", len(test_processed))
```

```
wer_metric = evaluate.load("wer")
cer_metric = evaluate.load("cer")
```

```
# 1. 4-bit Quantization Config
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

# 2. Force all layers to GPU 0 using device_map={"": 0}
model = WhisperForConditionalGeneration.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map={"": 0}
)
model.config.forced_decoder_ids = None
model.config.suppress_tokens = []
model.generation_config.language = "arabic"
model.generation_config.task = "transcribe"
model.eval()
print("Base model loaded successfully on GPU!")
```

```
def collate_fn(features):
    input_features = [{"input_features": f["input_features"]} for f in features]
    return processor.feature_extractor.pad(input_features, return_tensors="pt")

dataloader = DataLoader(test_processed, batch_size=8, collate_fn=collate_fn)

references = [
    processor.tokenizer.decode(lbl, skip_special_tokens=True)
    for lbl in test_processed["labels"]
]

predictions = []
with torch.no_grad():
    for batch in tqdm(dataloader, desc="Evaluating Base Model"):
        # Match both the model's device and float16 dtype
        inputs = batch["input_features"].to(device=model.device, dtype=torch.float16)
        pred_ids = model.generate(inputs, language="arabic", task="transcribe", max_length=128)
        preds = processor.batch_decode(pred_ids, skip_special_tokens=True)
        predictions.extend(preds)

base_wer = 100 * wer_metric.compute(predictions=predictions, references=references)
base_cer = 100 * cer_metric.compute(predictions=predictions, references=references)

print(f"\n--- Base Model Test Results ---")
print(f"WER: {base_wer:.2f}%")
print(f"CER: {base_cer:.2f}%")
```

```
# Quick test on a single sample (index 10)
sample = dataset["test"][10]
audio = sample["audio"]["array"]
ground_truth = sample["text"]

inputs = processor(audio, sampling_rate=16000, return_tensors="pt").input_features.half().cuda()
with torch.no_grad():
    pred_id = model.generate(inputs, language="arabic", task="transcribe")
pred_text = processor.batch_decode(pred_id, skip_special_tokens=True)[0]

print("Ground Truth:", ground_truth)
print("Prediction :", pred_text)
```

## Part 2 — LoRA Fine-tuning on Bahraini Data

This continues directly from Part 1 — it reuses `model`, `processor`, `train_processed`, `test_processed`, and `wer_metric` from above.

**Tip:** on first run, try `num_train_epochs=1` and a smaller subset of `train_processed` just to confirm everything runs before committing to a full multi-epoch run.

```
# ---- 1) PEFT / LoRA imports ----
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from transformers import Seq2SeqTrainingArguments, Seq2SeqTrainer
from dataclasses import dataclass
from typing import Any, Dict, List, Union
```

```
# ---- 2) Prepare the 4-bit model for training ----
model = prepare_model_for_kbit_training(model)
```

```
# ---- 3) LoRA configuration ----
# r / lora_alpha: higher values = more trainable parameters (and more memory)
# target_modules: Whisper's attention projections (standard choice)
lora_config = LoraConfig(
    r=32,
    lora_alpha=64,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none",
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
# Expect a small percentage (roughly 1-2%) of total parameters to be trainable
```

```
# ---- 4) Data collator ----
@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any

    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        input_features = [{"input_features": f["input_features"]} for f in features]
        batch = self.processor.feature_extractor.pad(input_features, return_tensors="pt")

        label_features = [{"input_ids": f["labels"]} for f in features]
        labels_batch = self.processor.tokenizer.pad(label_features, return_tensors="pt")

        # Replace padding with -100 so it's ignored in the loss
        labels = labels_batch["input_ids"].masked_fill(labels_batch.attention_mask.ne(1), -100)

        if (labels[:, 0] == self.processor.tokenizer.bos_token_id).all().cpu().item():
            labels = labels[:, 1:]

        batch["labels"] = labels
        return batch

data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor=processor)
```

```
# ---- 5) WER metric during training ----
def compute_metrics(pred):
    pred_ids = pred.predictions
    label_ids = pred.label_ids
    label_ids[label_ids == -100] = processor.tokenizer.pad_token_id

    pred_str = processor.batch_decode(pred_ids, skip_special_tokens=True)
    label_str = processor.batch_decode(label_ids, skip_special_tokens=True)

    wer = 100 * wer_metric.compute(predictions=pred_str, references=label_str)
    return {"wer": wer}
```

```
# ---- 6) Training arguments ----
# If time is tight, lower num_train_epochs to 1 or 2 to finish faster
training_args = Seq2SeqTrainingArguments(
    output_dir="./whisper-bahraini-lora",
    per_device_train_batch_size=8,
    gradient_accumulation_steps=2,
    learning_rate=1e-3,
    warmup_steps=50,
    num_train_epochs=3,
    gradient_checkpointing=True,
    fp16=True,
    per_device_eval_batch_size=8,
    predict_with_generate=True,
    generation_max_length=128,
    logging_steps=25,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    load_best_model_at_end=True,
    metric_for_best_model="wer",
    greater_is_better=False,
    report_to=["none"],
)

trainer = Seq2SeqTrainer(
    args=training_args,
    model=model,
    train_dataset=train_processed,
    eval_dataset=test_processed,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
    tokenizer=processor.feature_extractor,
)
```

```
# ---- 7) Train ----
model.config.use_cache = False  # required alongside gradient_checkpointing
trainer.train()
```

```
# ---- 8) Save the fine-tuned LoRA adapter ----
model.save_pretrained("./whisper-bahraini-lora-final")
processor.save_pretrained("./whisper-bahraini-lora-final")

print("\n--- Training complete and model saved ---")
print("Fine-tuned model saved to: ./whisper-bahraini-lora-final")
```
