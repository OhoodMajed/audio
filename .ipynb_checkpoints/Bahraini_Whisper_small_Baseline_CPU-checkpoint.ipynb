# Bahraini Model BaseLine Test

**Goal:** measure the stock `openai/whisper-small` on the Bahraini test clips

> Note: `whisper-small` (~244M params) is much lighter than `large-v3` (~1.5B params). 4-bit quantization isn't really necessary for a model this size, but it's kept here for consistency with the original workflow / GPU memory habits. Feel free to drop it and load in fp16 directly if you want a simpler, faster run.

## 3. Dataset Loading & Cleaning

```python
repo_id = "Hishambarakat/Bahraini_Speech_Dataset"
local_path = snapshot_download(
    repo_id=repo_id,
    repo_type="dataset",
    allow_patterns=["dataset_dict.json", "dataset_info.json", "train/**", "test/**"],
    max_workers=1,  # Forces sequential downloading to prevent Jupyter from freezing
    resume_download=True)  # Picks up where it left off if it drops

dataset = load_from_disk(local_path)
dataset = dataset.cast_column("audio", Audio(sampling_rate=16000))

print("Train clips:", len(dataset["train"]))
print("Test clips:", len(dataset["test"]))
```

```python
# Filter out empty text and audio clips shorter than 0.5s (8000 samples at 16kHz)
def is_valid(sample):
    valid_text = sample["text"] is not None and len(sample["text"].strip()) > 0
    valid_audio = len(sample["audio"]["array"]) >= 8000
    return valid_text and valid_audio
```

```python
dataset = dataset.filter(is_valid, num_proc=8)
print("Cleaned Train:", len(dataset["train"]))
print("Cleaned Test:", len(dataset["test"]))
```

## 4. Model & Processor Initialization

```python
model_id = "openai/whisper-small"
processor = WhisperProcessor.from_pretrained(
    model_id,
    language="Arabic",
    task="transcribe")
```

## 5. Test Data Preprocessing

```python
def prepare_dataset(batch, processor):
    audio = batch["audio"]
    # Extract audio features
    batch["input_features"] = processor.feature_extractor(audio["array"], sampling_rate=audio["sampling_rate"]).input_features[0]
    # Tokenize target text
    batch["labels"] = processor.tokenizer(batch["text"], truncation=True, max_length=448).input_ids
    return batch

# Select subsets: 5000 train, 500 test
train_sub = dataset["train"].select(range(5000))
test_sub = dataset["test"].select(range(500))
train_processed = train_sub.map(prepare_dataset, fn_kwargs={"processor": processor}, remove_columns=dataset["train"].column_names, num_proc=2)
test_processed = test_sub.map(prepare_dataset, fn_kwargs={"processor": processor}, remove_columns=dataset["test"].column_names, num_proc=2)
print("Train samples ready:", len(train_processed))
print("Test samples ready:", len(test_processed))
```

## 6. Baseline Evaluation (WER & CER)

```python
wer_metric = evaluate.load("wer")
cer_metric = evaluate.load("cer")
```

```python
dataset = dataset.filter(is_valid)

print("Cleaned Train:", len(dataset["train"]))
print("Cleaned Test:", len(dataset["test"]))
```

```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("CUDA version:", torch.version.cuda)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
# جهازنا بدون GPU، فنحمل الموديل عادي بدون quantization ونحطه على الـ CPU
model = WhisperForConditionalGeneration.from_pretrained(model_id)
model.to("cpu")
model.config.forced_decoder_ids = None
model.config.suppress_tokens = []
model.generation_config.language = "arabic"
model.generation_config.task = "transcribe"
model.eval()
print("Base model loaded successfully on CPU!")
```

```python
def collate_fn(features):
    input_features = [{"input_features": f["input_features"]} for f in features]
    return processor.feature_extractor.pad(input_features, return_tensors="pt")

dataloader = DataLoader(test_processed, batch_size=8, collate_fn=collate_fn)
references = [processor.tokenizer.decode(lbl, skip_special_tokens=True)
              for lbl in test_processed["labels"]
             ]
predictions = []
with torch.no_grad():
    for batch in tqdm(dataloader, desc="Evaluating Base Model"):
        # على الـ CPU نستخدم float32 عادي (float16 بطيء/غير مدعوم جيداً على CPU)
        inputs = batch["input_features"].to(device="cpu", dtype=torch.float32)
        pred_ids = model.generate(inputs, language="arabic", task="transcribe", max_length=128)
        preds = processor.batch_decode(pred_ids, skip_special_tokens=True)
        predictions.extend(preds)

base_wer = 100 * wer_metric.compute(predictions=predictions, references=references)
base_cer = 100 * cer_metric.compute(predictions=predictions, references=references)
print(f"\n--- Base Model Test Results---")
print(f"WER: {base_wer:.2f}%")
print(f"CER: {base_cer:.2f}%")
```

## 7. Single Sample Inference Test

```python
# Quick test on a single sample (index any)
sample = dataset["test"][9]
audio = sample["audio"]["array"]
ground_truth = sample["text"]
inputs = processor(audio, sampling_rate=16000, return_tensors="pt").input_features
with torch.no_grad():
    pred_id = model.generate(inputs, language="arabic", task="transcribe")
pred_text = processor.batch_decode(pred_id, skip_special_tokens=True)[0]
print("Truth:", ground_truth)
print("Prediction :", pred_text)
```
