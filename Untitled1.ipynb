```python

import os
import glob
import pandas as pd
from IPython.display import Audio, display, clear_output
from natsort import natsorted  # استيراد دالة الترتيب الطبيعي

# 1. جلب كل الملفات الصوتية وترتيبها ترتيباً رقمياً طبيعياً (1, 2, 3... 10)
raw_audio_files = glob.glob("*.wav")  # غيري الامتداد حسب ملفاتك
audio_files = natsorted(raw_audio_files)

data = []
csv_file = "metadata.csv"

# فحص الملف إذا كان موجوداً وغير فارغ
if os.path.exists(csv_file) and os.path.getsize(csv_file) > 0:
    try:
        df_existing = pd.read_csv(csv_file)
        done_files = set(df_existing["audio_file"])
        data = df_existing.to_dict('records')
    except Exception:
        done_files = set()
else:
    done_files = set()

# تصفية الملفات المتبقية مع الحفاظ على الترتيب الطبيعي
files_to_process = [f for f in audio_files if f not in done_files]

print(f"إجمالي الملفات: {len(audio_files)} | المتبقي للتفريغ: {len(files_to_process)}\n")
print("تعليمات:")
print("- اكتب النص واضغط Enter للتنسيق والانتقال للمقطع التالي.")
print("- اكتب 'r' لإعادة تشغيل المقطع.")
print("- اكتب 'exit' للحفظ والخروج.\n" + "="*40 + "\n")

for i, file_name in enumerate(files_to_process, 1):
    while True:
        print(f"[{i}/{len(files_to_process)}] جاري تشغيل: {file_name}")
        
        # عرض مشغل الصوت وتشغيله تلقائياً
        display(Audio(file_name, autoplay=True))
        
        text = input("التفريغ النصي: ").strip()
        
        if text.lower() == 'r':
            clear_output(wait=True)
            continue
        elif text.lower() == 'exit':
            break
        elif text == "":
            print("⚠️ لم تكتب شيئاً! اكتب النص أو اكتب 'r' لإعادة التشغيل.\n")
            continue
        else:
            data.append({"audio_file": file_name, "transcription": text})
            # حفظ تلقائي بعد كل مقطع
            pd.DataFrame(data).to_csv(csv_file, index=False, encoding="utf-8-sig")
            clear_output(wait=True)
            break
            
    if text.lower() == 'exit':
        print("\nتم إيقاف العمل وتأمين الحفظ بنجاح.")
        break

# حفظ نهائي
if data:
    df = pd.DataFrame(data)
    df.to_csv(csv_file, index=False, encoding="utf-8-sig")
    print(f"\n🎉 تم الانتهاء! تم حفظ {len(df)} مقطع في ملف: {csv_file}")
```
